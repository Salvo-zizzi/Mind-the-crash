### NA's Standardization 

```python

# Codici (oltre a -1) che la guida DfT 2024 indica come "sconosciuto / non assegnato / non definito".
# -1 = "Data missing or out of range" / "Unknown" / "Undefined" / "Code deprecated": vale per TUTTE le colonne.

MISSING, UNKNOWN = -1000, -2000
MISSING_LABEL, UNKNOWN_LABEL = "Data missing", "Unknown"
FLAG_LABELS = [MISSING_LABEL, UNKNOWN_LABEL]


extra_codes = {
    # ---------- COLLISION ----------
    "speed_limit": [99],
    "road_type": [9],
    "junction_detail": [99],
    "junction_detail_historic": [99],
    "junction_control": [9],
    "second_road_class": [9],
    "pedestrian_crossing": [99],
    "pedestrian_crossing_human_control_historic": [9],
    "pedestrian_crossing_physical_facilities_historic": [9],
    "weather_conditions": [9],
    "road_surface_conditions": [9],
    "special_conditions_at_site": [9],
    "carriageway_hazards": [99],
    "carriageway_hazards_historic": [9],
    "urban_or_rural_area": [3],                      # 3 = Unallocated

    # ---------- VEHICLE ----------
    "vehicle_type": [99],
    "towing_and_articulation": [9],
    "vehicle_manoeuvre": [99],
    "vehicle_manoeuvre_historic": [99],
    "vehicle_direction_from": [9],
    "vehicle_direction_to": [9],
    "vehicle_location_restricted_lane": [99],
    "vehicle_location_restricted_lane_historic": [99],
    "junction_location": [9],
    "skidding_and_overturning": [9],
    "hit_object_in_carriageway": [99],
    "vehicle_leaving_carriageway": [9],
    "hit_object_off_carriageway": [99],
    "first_point_of_impact": [9],
    "vehicle_left_hand_drive": [9],
    "journey_purpose_of_driver": [6],                # 6 = Not known or not requested
    "journey_purpose_of_driver_historic": [6, 15],   # 6 = Not known, 15 = Other/Not known
    "sex_of_driver": [3],                            # 3 = Not known

    # ---------- CASUALTY ----------
    "sex_of_casualty": [9],
    "casualty_type": [99],
    "pedestrian_location": [10],                     # 10 = Unknown or other
    "pedestrian_movement": [9],                      # 9 = Unknown or other
    "car_passenger": [9],
    "bus_or_coach_passenger": [9],
    "pedestrian_road_maintenance_worker": [2],       # 2 = Not known
}

def unknown_codes(col):
    return extra_codes.get(col, [])

def clean_missing(df, missing=MISSING, unknown=UNKNOWN):
    df = df.copy()
    for col in df.columns:
        s = df[col]
        if pd.api.types.is_numeric_dtype(s):
            s = s.mask(s.isin(unknown_codes(col)), unknown)   # 9, 99, 3... -> -2000
            s = s.mask(s == -1, missing)                      # -1 -> -1000
            s = s.fillna(missing)                             # NaN -> -1000
        else:
            s = s.mask(s.astype(str).str.strip().eq("-1"), str(missing))
            s = s.fillna(str(missing))
        df[col] = s
    return df
```

```python
"""Ricodifica dei codici numerici nelle etichette della guida DfT (2024 code list).
I codici "sconosciuto / mancante" (-1, 9, 99...) NON sono nei dizionari: recode() li trasforma in "-1000"."""


# ==================== COLLISION ====================
labels_collision = {
    "collision_severity": {
        1: "Fatal",
        2: "Serious",
        3: "Slight",
    },
    "day_of_week": {
        1: "Sunday",
        2: "Monday",
        3: "Tuesday",
        4: "Wednesday",
        5: "Thursday",
        6: "Friday",
        7: "Saturday",
    },
    "road_type": {
        1: "Roundabout",
        2: "One way street",
        3: "Dual carriageway",
        6: "Single carriageway",
        7: "Slip road",
        12: "One way street/Slip road",
    },
    "junction_control": {
        0: "Not at junction or within 20 metres",
        1: "Authorised person",
        2: "Auto traffic signal",
        3: "Stop sign",
        4: "Give way or uncontrolled",
    },
    "pedestrian_crossing": {
        0: "No physical crossing facility within 50m",
        11: "Human crossing control by school crossing patrol",
        12: "Human crossing control by other authorised person",
        13: "Zebra crossing",
        14: "Pedestrian light crossing (pelican or puffin or toucan or similar)",
        15: "Pedestrian phase at traffic signal",
        16: "Footbridge or subway",
        17: "Central refuge - no other controls",
    },
    "light_conditions": {
        1: "Daylight",
        4: "Darkness - lights lit",
        5: "Darkness - lights unlit",
        6: "Darkness - no lighting",
        7: "Darkness - lighting unknown",
    },
    "weather_conditions": {
        1: "Fine no high winds",
        2: "Raining no high winds",
        3: "Snowing no high winds",
        4: "Fine + high winds",
        5: "Raining + high winds",
        6: "Snowing + high winds",
        7: "Fog or mist",
        8: "Other",
    },
    "road_surface_conditions": {
        1: "Dry",
        2: "Wet or damp",
        3: "Snow",
        4: "Frost or ice",
        5: "Flood over 3cm. deep",
        6: "Oil or diesel",
        7: "Mud",
    },
    "special_conditions_at_site": {
        0: "None",
        1: "Auto traffic signal - out",
        2: "Auto signal part defective",
        3: "Road sign or marking defective or obscured",
        4: "Roadworks",
        5: "Road surface defective",
        6: "Oil or diesel",
        7: "Mud",
    },
    "urban_or_rural_area": {
        1: "Urban",
        2: "Rural",
    },
    "did_police_officer_attend_scene_of_accident": {
        1: "Yes",
        2: "No",
        3: "No - accident was reported using a self completion  form (self rep only)",
    },
    "trunk_road_flag": {
        1: "Trunk (Roads managed by Highways England)",
        2: "Non-trunk",
    },
}


# ==================== VEHICLE ====================
labels_vehicle = {
    "vehicle_type": {
        1: "Pedal cycle",
        2: "Motorcycle 50cc and under",
        3: "Motorcycle 125cc and under",
        4: "Motorcycle over 125cc and up to 500cc",
        5: "Motorcycle over 500cc",
        8: "Taxi/Private hire car",
        9: "Car",
        10: "Minibus (8 - 16 passenger seats)",
        11: "Bus or coach (17 or more pass seats)",
        16: "Ridden horse",
        17: "Agricultural vehicle",
        18: "Tram",
        19: "Van / Goods 3.5 tonnes mgw or under",
        20: "Goods over 3.5t. and under 7.5t",
        21: "Goods 7.5 tonnes mgw and over",
        22: "Mobility scooter",
        23: "Electric motorcycle",
        33: "Personal powered transporter",
        90: "Other vehicle",
        97: "Motorcycle - unknown cc",
        98: "Goods vehicle - unknown weight",
        103: "Motorcycle - Scooter (1979-1998)",
        104: "Motorcycle (1979-1998)",
        105: "Motorcycle - Combination (1979-1998)",
        106: "Motorcycle over 125cc (1999-2004)",
        108: "Taxi (excluding private hire cars) (1979-2004)",
        109: "Car (including private hire cars) (1979-2004)",
        110: "Minibus/Motor caravan (1979-1998)",
        113: "Goods over 3.5 tonnes (1979-1998)",
    },
    "towing_and_articulation": {
        0: "No tow/articulation",
        1: "Articulated vehicle",
        2: "Double or multiple trailer",
        3: "Caravan",
        4: "Single trailer",
        5: "Other tow",
    },
    "vehicle_manoeuvre_historic": {
        1: "Reversing",
        2: "Parked",
        3: "Waiting to go - held up",
        4: "Slowing or stopping",
        5: "Moving off",
        6: "U-turn",
        7: "Turning left",
        8: "Waiting to turn left",
        9: "Turning right",
        10: "Waiting to turn right",
        11: "Changing lane to left",
        12: "Changing lane to right",
        13: "Overtaking moving vehicle - offside",
        14: "Overtaking static vehicle - offside",
        15: "Overtaking - nearside",
        16: "Going ahead left-hand bend",
        17: "Going ahead right-hand bend",
        18: "Going ahead other",
    },
    "vehicle_manoeuvre": {
        1: "Reversing",
        2: "Parked",
        3: "Waiting to go ahead",
        4: "Slowing or stopping",
        5: "Moving off",
        6: "U-turn",
        7: "Turning left",
        8: "Waiting to turn left",
        9: "Turning right",
        10: "Waiting to turn right",
        11: "Changing lane to left",
        12: "Changing lane to right",
        13: "Over taking moving vehicle on its offside",
        14: "Overtaking stationary vehicle on its offside",
        15: "Overtaking on nearside (passengers side nearest kerb)",
        19: "Going ahead",
        20: "Parking",
    },
    "vehicle_location_restricted_lane_historic": {
        0: "On main c'way - not in restricted lane",
        1: "Tram/Light rail track",
        2: "Bus lane",
        3: "Busway (including guided busway)",
        4: "Cycle lane (on main carriageway)",
        5: "Cycleway or shared use footway (not part of  main carriageway)",
        6: "On lay-by or hard shoulder",
        7: "Entering lay-by or hard shoulder",
        8: "Leaving lay-by or hard shoulder",
        9: "Footway (pavement)",
        10: "Not on carriageway",
    },
    "vehicle_location_restricted_lane": {
        0: "On main carriageway (not in restricted lane)",
        1: "Tram or Light rail track",
        2: "Bus lane or Busway",
        4: "Cycle lane (on main carriageway)",
        5: "Cycleway or shared use footway (not part of main carriageway)",
        6: "Lay-by or hard shoulder",
        9: "Footway (pavement)",
    },
    "junction_location": {
        0: "Not at or within 20 metres of junction",
        1: "Approaching junction or waiting/parked at junction approach",
        2: "Cleared junction or waiting/parked at junction exit",
        3: "Leaving roundabout",
        4: "Entering roundabout",
        5: "Leaving main road",
        6: "Entering main road",
        7: "Entering from slip road",
        8: "Mid Junction - on roundabout or on main road",
    },
    "skidding_and_overturning": {
        0: "None",
        1: "Skidded",
        2: "Skidded and overturned",
        3: "Jackknifed",
        4: "Jackknifed and overturned",
        5: "Overturned",
    },
    "hit_object_in_carriageway": {
        0: "None",
        1: "Previous accident",
        2: "Road works",
        4: "Parked vehicle",
        5: "Bridge (roof)",
        6: "Bridge (side)",
        7: "Bollard or refuge",
        8: "Open door of vehicle",
        9: "Central island of roundabout",
        10: "Kerb",
        11: "Other object",
        12: "Any animal (except ridden horse)",
    },
    "vehicle_leaving_carriageway": {
        0: "Did not leave carriageway",
        1: "Nearside",
        2: "Nearside and rebounded",
        3: "Straight ahead at junction",
        4: "Offside on to central reservation",
        5: "Offside on to centrl res + rebounded",
        6: "Offside - crossed central reservation",
        7: "Offside",
        8: "Offside and rebounded",
    },
    "hit_object_off_carriageway": {
        0: "None",
        1: "Road sign or traffic signal",
        2: "Lamp post",
        3: "Telegraph or electricity pole",
        4: "Tree",
        5: "Bus stop or bus shelter",
        6: "Central crash barrier",
        7: "Near/Offside crash barrier",
        8: "Submerged in water",
        9: "Entered ditch",
        10: "Other permanent object",
        11: "Wall or fence",
    },
    "first_point_of_impact": {
        0: "Did not impact",
        1: "Front",
        2: "Back",
        3: "Offside",
        4: "Nearside",
    },
    "vehicle_left_hand_drive": {
        1: "No",
        2: "Yes",
    },
    "journey_purpose_of_driver_historic": {
        1: "Journey as part of work",
        2: "Commuting to/from work",
        3: "Taking pupil to/from school",
        4: "Pupil riding to/from school",
        5: "Other",
    },
    "journey_purpose_of_driver": {
        1: "Journey as part of work",
        2: "Commuting to or from work",
        7: "Education and educational escort",
        8: "Emergency vehicle (blue light) on response",
        9: "Personal business or leisure",
    },
    "sex_of_driver": {
        1: "Male",
        2: "Female",
    },
    "age_band_of_driver": {
        1: "0 - 5",
        2: "6 - 10",
        3: "11 - 15",
        4: "16 - 20",
        5: "21 - 25",
        6: "26 - 35",
        7: "36 - 45",
        8: "46 - 55",
        9: "56 - 65",
        10: "66 - 75",
        11: "Over 75",
    },
    "propulsion_code": {
        1: "Petrol",
        2: "Heavy oil",
        3: "Electric",
        4: "Steam",
        5: "Gas",
        6: "Petrol/Gas (LPG)",
        7: "Gas/Bi-fuel",
        8: "Hybrid electric",
        9: "Gas Diesel",
        10: "New fuel technology",
        11: "Fuel cells",
        12: "Electric diesel",
    },
    "driver_imd_decile": {
        1: "Most deprived 10%",
        2: "More deprived 10-20%",
        3: "More deprived 20-30%",
        4: "More deprived 30-40%",
        5: "More deprived 40-50%",
        6: "Less deprived 40-50%",
        7: "Less deprived 30-40%",
        8: "Less deprived 20-30%",
        9: "Less deprived 10-20%",
        10: "Least deprived 10%",
    },
    "escooter_flag": {
        1: "Vehicle was an e-scooter",
        0: "Vehicle was not an e-scooter",
    },
}


# ==================== CASUALTY ====================
labels_casualty = {
    "casualty_class": {
        1: "Driver or rider",
        2: "Passenger",
        3: "Pedestrian",
    },
    "pedestrian_location": {
        0: "Not a Pedestrian",
        1: "Crossing on pedestrian crossing facility",
        2: "Crossing in zig-zag approach lines",
        3: "Crossing in zig-zag exit lines",
        4: "Crossing elsewhere within 50m. of pedestrian crossing",
        5: "In carriageway, crossing elsewhere",
        6: "On footway or verge",
        7: "On refuge, central island or central reservation",
        8: "In centre of carriageway - not on refuge, island or central reservation",
        9: "In carriageway, not crossing",
    },
    "pedestrian_movement": {
        0: "Not a Pedestrian",
        1: "Crossing from driver's nearside",
        2: "Crossing from nearside - masked by parked or stationary vehicle",
        3: "Crossing from driver's offside",
        4: "Crossing from offside - masked by  parked or stationary vehicle",
        5: "In carriageway, stationary - not crossing  (standing or playing)",
        6: "In carriageway, stationary - not crossing  (standing or playing) - masked by parked or stationary vehicle",
        7: "Walking along in carriageway, facing traffic",
        8: "Walking along in carriageway, back to traffic",
    },
    "casualty_type": {
        0: "Pedestrian",
        1: "Cyclist",
        2: "Motorcycle 50cc and under rider or passenger",
        3: "Motorcycle 125cc and under rider or passenger",
        4: "Motorcycle over 125cc and up to 500cc rider or  passenger",
        5: "Motorcycle over 500cc rider or passenger",
        8: "Taxi/Private hire car occupant",
        9: "Car occupant",
        10: "Minibus (8 - 16 passenger seats) occupant",
        11: "Bus or coach occupant (17 or more pass seats)",
        16: "Horse rider",
        17: "Agricultural vehicle occupant",
        18: "Tram occupant",
        19: "Van / Goods vehicle (3.5 tonnes mgw or under) occupant",
        20: "Goods vehicle (over 3.5t. and under 7.5t.) occupant",
        21: "Goods vehicle (7.5 tonnes mgw and over) occupant",
        22: "Mobility scooter rider",
        23: "Electric motorcycle rider or passenger",
        33: "Personal Powered Transporter",
        90: "Other vehicle occupant",
        97: "Motorcycle - unknown cc rider or passenger",
        98: "Goods vehicle (unknown weight) occupant",
        103: "Motorcycle - Scooter (1979-1998)",
        104: "Motorcycle (1979-1998)",
        105: "Motorcycle - Combination (1979-1998)",
        106: "Motorcycle over 125cc (1999-2004)",
        108: "Taxi (excluding private hire cars) (1979-2004)",
        109: "Car (including private hire cars) (1979-2004)",
        110: "Minibus/Motor caravan (1979-1998)",
        113: "Goods over 3.5 tonnes (1979-1998)",
    },
}


# ==================== COLLISION: region e road_class_name ====================
police_force_to_region = {
    # London
    1: 'London', 48: 'London',
    # North East
    10: 'North East', 11: 'North East', 17: 'North East',
    # North West
    3: 'North West', 4: 'North West', 5: 'North West', 6: 'North West', 7: 'North West',
    # Yorkshire and the Humber
    12: 'Yorkshire and the Humber', 13: 'Yorkshire and the Humber',
    14: 'Yorkshire and the Humber', 16: 'Yorkshire and the Humber',
    # West Midlands
    20: 'West Midlands', 21: 'West Midlands', 22: 'West Midlands', 23: 'West Midlands',
    # East Midlands
    30: 'East Midlands', 31: 'East Midlands', 32: 'East Midlands',
    33: 'East Midlands', 34: 'East Midlands',
    # East of England
    35: 'East of England', 36: 'East of England', 37: 'East of England',
    40: 'East of England', 41: 'East of England', 42: 'East of England',
    # South East
    43: 'South East', 44: 'South East', 45: 'South East', 46: 'South East', 47: 'South East',
    # South West
    50: 'South West', 52: 'South West', 53: 'South West', 54: 'South West', 55: 'South West',
    # Wales
    60: 'Wales', 61: 'Wales', 62: 'Wales', 63: 'Wales',
    # Scotland
    91: 'Scotland', 92: 'Scotland', 93: 'Scotland', 94: 'Scotland', 95: 'Scotland',
    96: 'Scotland', 97: 'Scotland', 98: 'Scotland', 99: 'Scotland',
}

road_class_short = {
    1: "M",      # Motorway / Autostrada
    2: "A(M)",   # Strada A a standard autostradale
    3: "A",      # Strada principale
    4: "B",      # Strada secondaria
    5: "C",      # Strada locale minore
    6: "U",      # Unclassified / Strada locale
}

road_class_name = {
    "M": "Motorway",
    "A(M)": "Motorway-standard A road",
    "A": "Main road",
    "B": "Secondary road",
    "C": "Minor local road",
    "U": "Local road",
}


def _map_with_flags(s, mapping, missing_label, unknown_label):
    out = s.map(mapping)
    unmapped = s[out.isna() & ~s.isin([MISSING, UNKNOWN])].unique()
    assert len(unmapped) == 0, f"{s.name}: codici non mappati {sorted(unmapped)}"
    out = out.mask(s == UNKNOWN, unknown_label)
    out = out.mask(s == MISSING, missing_label)
    return out


def recode(df, labels, missing_label=MISSING_LABEL, unknown_label=UNKNOWN_LABEL):
    df = df.copy()
    for col, mapping in labels.items():
        if col in df.columns:
            df[col] = _map_with_flags(df[col], mapping, missing_label, unknown_label)
    return df


def recode_collision(df, missing_label=MISSING_LABEL, unknown_label=UNKNOWN_LABEL):
    df = df.copy()

    # first_road_class -> road_class_name
    rc = {k: road_class_name[v] for k, v in road_class_short.items()}
    df["road_class_name"] = _map_with_flags(df["first_road_class"], rc, missing_label, unknown_label)
    df = df.drop(columns="first_road_class")

    # tutte le altre colonne con etichette
    return recode(df, labels_collision, missing_label, unknown_label)
```

## Funzioni di encoding

```python
def make_flag(df, col, zero_values):
    """0 se il valore è in `zero_values`, 1 altrimenti.
    Va usata DOPO recode(): `zero_values` sono quindi etichette di testo (inclusi i mancanti)."""
    df = df.copy()
    df[col] = (~df[col].isin(zero_values)).astype("int8")
    return df

def _rename_flags(cols):
    """'x_Data missing' -> 'x_missing',  'x_Unknown' -> 'x_unknown'"""
    ren = {}
    for c in cols:
        if c.endswith("_" + MISSING_LABEL):
            ren[c] = c[:-len(MISSING_LABEL)] + "missing"
        elif c.endswith("_" + UNKNOWN_LABEL):
            ren[c] = c[:-len(UNKNOWN_LABEL)] + "unknown"
    return ren

def one_hot(df, cols, drop=(), drop_first=False, dtype="int8"):
    """drop: flag da eliminare, es. ("missing",) oppure ("missing", "unknown")."""
    df = df.copy()
    cols = list(dict.fromkeys(cols))
    d = pd.get_dummies(df[cols].astype(str), prefix=cols, drop_first=drop_first, dtype=dtype)
    d = d.rename(columns=_rename_flags(d.columns))
    d = d.drop(columns=[c for c in d.columns if c.endswith(tuple("_" + s for s in drop))])
    return pd.concat([df.drop(columns=cols), d], axis=1)

def fill_from_historic(df, col, hist_col, mapping):
    df = df.copy()
    mask = df[col] == MISSING_LABEL
    hist = df.loc[mask, hist_col]
    filled = hist.map(mapping).where(~hist.isin(FLAG_LABELS), hist)
    df.loc[mask, col] = filled.fillna(MISSING_LABEL)
    return df.drop(columns=hist_col)


def time_to_period(df, col="time", missing=MISSING_LABEL):
    df = df.copy()
    hour = pd.to_numeric(df[col].astype(str).str[:2], errors="coerce")
    df[col + "_period"] = pd.cut(hour, bins=[0, 6, 12, 18, 24], right=False,
                                 labels=["night", "morning", "afternoon", "evening"]
                                 ).astype(object).fillna(missing).astype(str)
    return df.drop(columns=col)


def aggregate_counts(df, id_col="collision_index", mean_cols=()):
    mean_cols = list(mean_cols)
    sum_cols = [c for c in df.columns if c != id_col and c not in mean_cols]
    out = df.groupby(id_col)[sum_cols].sum().astype("int16")
    for c in mean_cols:
        out[f"{c}_mean"] = df[c].where(~df[c].isin([MISSING, UNKNOWN])).groupby(df[id_col]).mean()
    return out
```

```python
import pandas as pd

CORR_THR, MATCH_THR = 0.8, 0.999
log = []
gc = lambda d, p: [c for c in d.columns if c.startswith(p)]
res = lambda d, n: n if n in d.columns else next(iter([c for c in d.columns if c.startswith(n)]), None)

# --- 1. Ridondanze certe + regione (ho lat/lon) ---
certain = ["escooter_flag_Vehicle was not an e-scooter", "trunk_road_flag_Non-trunk",
           "vehicle_left_hand_drive_No", "pedestrian_movement_Not a Pedestrian"]
cols = [c for c in certain if c in df.columns] + gc(df, "region_")
log += [(c, "certa") for c in cols]
df_clean = df.drop(columns=cols)

# --- 2. Dummy trap (one-hot livello incidente) ---
print("=== DUMMY TRAP ===")
for p in ["day_of_week_", "junction_control_", "light_conditions_", "pedestrian_crossing_",
          "road_class_name_", "road_surface_conditions_", "road_type_", "time_period_",
          "trunk_road_flag_", "urban_or_rural_area_", "weather_conditions_"]:
    cols = gc(df_clean, p)
    if len(cols) < 2: continue
    s = df_clean[cols].sum(axis=1)
    print(f"{p:<28} sum==1: {(s==1).mean():.4f}  sum==0: {(s==0).mean():.4f}  sum>1: {(s>1).mean():.4f}")
    if (s == 1).mean() >= MATCH_THR:
        ref = df_clean[cols].sum().idxmax()
        df_clean = df_clean.drop(columns=ref); log.append((ref, f"dummy trap {p}"))

# --- 3. Conteggi aggregati ---
print("\n=== CONTEGGI ===")
count_groups = {
    "vehicle_type_": "number_of_vehicles", "vehicle_manoeuvre_": "number_of_vehicles",
    "vehicle_location_restricted_lane_": "number_of_vehicles", "junction_location_": "number_of_vehicles",
    "skidding_and_overturning_": "number_of_vehicles", "first_point_of_impact_": "number_of_vehicles",
    "sex_of_driver_": "number_of_vehicles", "age_band_of_driver_": "number_of_vehicles",
    "casualty_class_": "number_of_casualties", "casualty_type_": "number_of_casualties",
    "pedestrian_location_": "casualty_class_Pedestrian", "pedestrian_movement_": "casualty_class_Pedestrian",
}
for p, tot in count_groups.items():
    cols = gc(df_clean, p)
    if len(cols) < 2 or tot not in df_clean.columns: continue
    s, t = df_clean[cols].sum(axis=1), df_clean[tot]
    cap = t.max() if (t.nunique() <= 5 and tot.startswith("number_of")) else None
    sc = s.clip(upper=cap) if cap is not None else s
    eq, gt = (sc == t).mean(), (sc > t).mean()
    print(f"{p:<36} vs {tot:<26} ==: {eq:.4f}  >: {gt:.4f}" + (f"  (cap={cap})" if cap else ""))
    if gt > 0: print("   ! somma > totale: possibile inflazione da merge veicoli x vittime")
    if eq >= MATCH_THR:
        ref = df_clean[cols].sum().idxmax()
        df_clean = df_clean.drop(columns=ref); log.append((ref, f"conteggio = {tot}"))

# --- 4. Ridondanze semantiche ---
print("\n=== SEMANTICHE ===")
for keep, drop in [("junction_control_Not at junction", "junction_location_Not at junction"),
                   ("road_class_name_Motorway", "trunk_road_flag_Trunk"),
                   ("vehicle_type_Car", "casualty_type_Car"),
                   ("vehicle_type_Bus or minibus", "casualty_type_Bus or minibus"),
                   ("vehicle_type_Van", "casualty_type_Van")]:
    k, d = res(df_clean, keep), res(df_clean, drop)
    if not (k and d): print(f"{keep} / {drop}: mancante, salto"); continue
    r = df_clean[k].corr(df_clean[d])
    print(f"corr({k}, {d}) = {r:+.3f}" + ("  -> TOLGO" if abs(r) >= CORR_THR else ""))
```
